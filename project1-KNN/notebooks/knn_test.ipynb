```python
import pandas as pd
import numpy as np

import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.tuning import tune_knn_classification_params, tune_knn_regression_params
from src.evaluation import cross_validation_loop
```

```python
# Breast Cancer Dataset - Classification
tuning_iterations = 100
breast_cancer = pd.read_csv("../data/processed/breast_cancer_processed.csv")

feature_data = breast_cancer.drop(columns=["class"]).to_numpy()
label_data = breast_cancer["class"].to_numpy()

k = tune_knn_classification_params(feature_data, label_data, n=tuning_iterations)[0]
print(f"Tuned k for Breast Cancer dataset: {k}")

classification_error_breast_cancer =cross_validation_loop(feature_data, label_data, k=k, model='classification')
print("Classification Error for Breast Cancer Dataset:", classification_error_breast_cancer)
print("Classification Eror Average for Breast Cancer Dataset:", format(np.mean(classification_error_breast_cancer), ".4f"))
```

```python
# Abalone Dataset - Regression
tuning_iterations = 250
abalone = pd.read_csv("../data/processed/abalone_processed.csv")

feature_data = abalone.drop(columns=["rings"]).to_numpy()
label_data = abalone["rings"].to_numpy()

k, gamma = tune_knn_regression_params(feature_data, label_data, n=tuning_iterations)[:2]
print(f"Tuned k for Abalone dataset: {k}")
print(f"Tuned gamma for edited Abalone dataset: {gamma}")

regression_error_abalone =cross_validation_loop(feature_data, label_data, k=k, model='regression', gamma=gamma)
print("Mean Squared Error for Abalone Dataset:", regression_error_abalone)
print("Mean Squared Error Average for Abalone Dataset:", format(np.mean(regression_error_abalone), ".4f"))
```
