```python
import pandas as pd
import numpy as np

import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.reduction import edited_knn, condensed_knn
from src.evaluation import cross_validation_loop
from src.tuning import tune_knn_classification_params, tune_knn_regression_params, tune_epsilon_param
from src.null_model import null_model_classification, null_model_regression
```

#### Breast Cancer Dataset - Classification
iterations = 5 ran for 18.7s

```python
# Breast Cancer Dataset - Classification
tuning_iterations = 5
breast_cancer = pd.read_csv("../data/processed/breast_cancer_processed.csv")

feature_data = breast_cancer.drop(columns=["class"])
label_data = breast_cancer["class"]

null_model_classification_error = null_model_classification(label_data)
print(f"Null model classification error for Breast Cancer dataset: {format(null_model_classification_error, '.4f')}\n")

edited_features, edited_labels = edited_knn(feature_data, label_data, task="classification")
edited_k = tune_knn_classification_params(edited_features, edited_labels, n=tuning_iterations)[0]
print(f"Tuned k for edited Breast Cancer dataset: {edited_k}")
classification_error_edited = np.mean(cross_validation_loop(edited_features, edited_labels, k=edited_k, model='classification')) # type: ignore
print(f"Classification error for Breast Cancer dataset after Edited KNN: {format(classification_error_edited, '.4f')}\n")

condensed_features, condensed_labels = condensed_knn(feature_data, label_data, task="classification")
condensed_k = tune_knn_classification_params(condensed_features, condensed_labels, n=tuning_iterations)[0]
print(f"Tuned k for condensed Breast Cancer dataset: {condensed_k}")
classification_error_condensed = np.mean(cross_validation_loop(condensed_features, condensed_labels, k=condensed_k, model='classification')) # type: ignore
print(f"Classification error for Breast Cancer dataset after Condensed KNN: {format(classification_error_condensed, '.4f')}")
```
