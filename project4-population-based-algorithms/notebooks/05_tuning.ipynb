# 05 - Tune Hyperparameters

```python
import pandas as pd
import numpy as np

import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.preprocessing import load_processed, prepare_xy
from src.neural_network_population_trained import NeuralNetwork
from src.evaluation import cross_validation_loop
from src.differential_evolution import differential_evolution
from src.genetic_algorithm import genetic_algorithm
from src.tuning import holdout_split, random_search
```

```python
seed = 0
```

### Breast Cancer Dataset

```python
data, meta = load_processed("../data/processed/breast_cancer_processed.csv")
X, y, y_labels = prepare_xy(data, meta)

# 80/20 train/validation split
train_idx, val_idx = holdout_split(len(X), y_labels, seed=seed)

network = NeuralNetwork([X.shape[1], 15, 15, y.shape[1]], mode="classification", seed=seed)
```

__Genetic Algorithm__

```python
# Hyperparameter tuning
best_params, best_score, trials = random_search(
    20,                                         # n_trials
    X[train_idx], y[train_idx],                 
    X[val_idx], y[val_idx], y_labels[val_idx],  
    network, genetic_algorithm,
    mode="classification", seed=seed,
)
print("Best params:", best_params)

error = cross_validation_loop(X, y, y_labels, network, best_params, algorithm=genetic_algorithm, mode="classification", seed=seed)
print(f"Mean Classification Error: {np.mean(error['error']):.4f}")
```

__Differential Evolution__

```python
# Hyperparameter tuning
best_params, best_score, trials = random_search(
    20,                                         # n_trials
    X[train_idx], y[train_idx],                 
    X[val_idx], y[val_idx], y_labels[val_idx],  
    network, differential_evolution,
    mode="classification", seed=seed,
)
print("Best params:", best_params)

error = cross_validation_loop(X, y, y_labels, network, best_params, algorithm=differential_evolution, mode="classification", seed=seed)
print(f"Mean Classification Error: {np.mean(error['error']):.4f}")
```
