```python
import pandas as pd
import numpy as np

import sys
from pathlib import Path

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.decision_tree import DecisionTree
from src.preprocessing import load_processed
from src.evaluation import cross_validation_loop
```

```python
breast_cancer, feature_types = load_processed("../data/processed/breast_cancer_processed.csv")

feature_data = breast_cancer.drop(columns=["class"]).to_numpy()
label_data = breast_cancer["class"].to_numpy()

results = cross_validation_loop(feature_data, label_data, feature_types, mode="classification", seed=0)

print(f"Classification Error: {format(np.mean(results['unpruned']), '.4f')}")
print(f"Pruned Error: {format(np.mean(results['pruned']), '.4f')}\n")

print(f"Mean Unpruned Tree Size: {format(np.mean(results['unpruned size']), '.2f')}")
print(f"Mean Pruned Tree Size: {format(np.mean(results['pruned size']), '.2f')}")
```
